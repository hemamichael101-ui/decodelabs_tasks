```python
def analyze_message(message_text):
    """Analyzes a message text for common phishing red flags and keywords."""
    print("\n================ THREAT ANALYSIS REPORT ================")

    # 1. Define common phishing red flags/keywords
    urgency_keywords = ["urgent", "immediate action", "suspicious activity", "locked", "unauthorized login", "act now", "final notice"]
    financial_keywords = ["bank", "crypto", "verify account", "billing", "payment details", "invoice", "refund"]
    suspicious_links = [".net/secure", "bit.ly/", "login-update-verify", "secure-bank-login", ".xyz", "http://"]

    red_flags_found = []
    unsafe_reasons = []

    # 2. Check for Urgency Keywords
    found_urgency = [word for word in urgency_keywords if word in message_text.lower()]
    if found_urgency:
        red_flags_found.append(f"Urgent Language: {found_urgency}")
        unsafe_reasons.append("Creating a false sense of urgency forces victims to act quickly without thinking.")

    # 3. Check for Financial/Account Verification Keywords
    found_financial = [word for word in financial_keywords if word in message_text.lower()]
    if found_financial:
        red_flags_found.append(f"Financial/Credential Request: {found_financial}")
        unsafe_reasons.append("Phishing emails frequently ask you to update passwords, billing details, or verify accounts.")

    # 4. Check for Suspicious Links
    found_links = [link for link in suspicious_links if link in message_text.lower()]
    if found_links:
        red_flags_found.append(f"Suspicious URL/Domain Indicators: {found_links}")
        unsafe_reasons.append("The message contains unverified links or non-secure (HTTP) addresses designed to mimic real login pages.")

    # 5. Display the Results
    if red_flags_found:
        print("🚨 STATUS: SUSPICIOUS / POTENTIAL PHISHING ATTEMPT 🚨\n")
        print("--- Red Flags Found ---")
        for flag in red_flags_found:
            print(f"- {flag}")

        print("\n--- Why This Message is Unsafe ---")
        for reason in unsafe_reasons:
            print(f"👉 {reason}")
    else:
        print("✅ STATUS: NO COMMON PHISHING INDICATORS FOUND")
        print("Note: Always stay cautious even if automated filters pass the message.")
    print("========================================================")


def main():
    print("--- Project 3: Phishing Awareness Analyzer ---")

    # Pre-loaded sample emails for testing
    samples = {
        "1": "URGENT: Your Bank account has experienced an unauthorized login. Please click http://secure-bank-login.xyz to verify your payment details immediately or your account will be permanently locked.",
        "2": "Hi Team, just a reminder that our general monthly sync meeting is scheduled for tomorrow at 10 AM in Conference Room B. See you there!",
        "3": "Final Notice: Your netflix billing information has failed. Act now to prevent service interruption. Click bit.ly/update-pay to fix this."
    }

    print("\nChoose an option:")
    print("1 - Test Sample Email 1 (Bank Alert)")
    print("2 - Test Sample Email 2 (Work Meeting Reminder)")
    print("3 - Test Sample Email 3 (Subscription Update)")
    print("4 - Paste your own custom message")

    choice = input("\nEnter your choice (1-4): ")

    if choice in ["1", "2", "3"]:
        selected_text = samples[choice]
        print(f"\nAnalyzing Sample Message:\n\"{selected_text}\"")
        analyze_message(selected_text)
    elif choice == "4":
        custom_text = input("\nPaste or type the suspicious message here:\n")
        analyze_message(custom_text)
    else:
        print("Invalid choice. Please run the program again.")

if __name__ == "__main__":
    main()
```
