```python
def caesar_encrypt(text, shift):
    encrypted_text = ""
    for char in text:
        if char.isalpha():
            start = ord('A') if char.isupper() else ord('a')
            new_char = chr(start + (ord(char) - start + shift) % 26)
            encrypted_text += new_char
        else:
            encrypted_text += char
    return encrypted_text

def caesar_decrypt(text, shift):
    return caesar_encrypt(text, -shift)

def main():
    print("--- Project 2: Encryption & Decryption System ---")
    user_text = input("Enter your message: ")
    shift_key = int(input("Enter a shift key number (e.g., 3): "))

    encrypted_msg = caesar_encrypt(user_text, shift_key)
    decrypted_msg = caesar_decrypt(encrypted_msg, shift_key)

    print("\n--- RESULTS ---")
    print(f"Original Text:  {user_text}")
    print(f"Encrypted Text: {encrypted_msg}")
    print(f"Decrypted Text: {decrypted_msg}")

if __name__ == "__main__":
    main()
```
